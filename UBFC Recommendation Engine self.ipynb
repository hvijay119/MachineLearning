### 4.1  Building UBCF using Pearson CorrelationMethod

```python
movies_data.head(2)
```

```python
pivot_table=pd.pivot_table(data=movies_data,values="rating",index="movie",columns="userId").fillna(0)
pivot_table.columns=movies_data["userId"].unique()
pivot_table
```

```python
# method: 'CorrelationMethod' = 'pearson', values between -1 to +1
pivot_table.corr().round(3)
```

### 4.2. Building UBCF using Euclideon Distance

```python
from sklearn.metrics import pairwise_distances
```

```python
movies_data.head(5)
```

```python
euclideon_pivot_table=pd.pivot_table(data=movies_data,values="rating",index="userId",columns="movie").fillna(0)
euclideon_pivot_table.columns=movies_data["movie"].unique()
euclideon_pivot_table.index=movies_data["userId"].unique()
euclideon_pivot_table
```

```python
eculideon_correlation=pd.DataFrame(pairwise_distances(X=euclideon_pivot_table,metric='euclidean'))
eculideon_correlation.index=movies_data["userId"].unique()
eculideon_correlation.columns=movies_data["userId"].unique()
eculideon_correlation
```

## 4.3. Building UBCF using Cosine Similarity

```python
eculideon_correlation=pd.DataFrame(pairwise_distances(X=euclideon_pivot_table,metric='cosine'))
eculideon_correlation.index=movies_data["userId"].unique()
eculideon_correlation.columns=movies_data["userId"].unique()
eculideon_correlation
```
