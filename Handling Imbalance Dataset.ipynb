## 1. Import Necessary Libraries

```python
import pandas as pd
from sklearn import datasets
from matplotlib import pyplot as plt
from sklearn.model_selection import train_test_split
```

## 2. Import Dataset

```python
breast_cancer_data=datasets.load_breast_cancer()
```

```python
breast_cancer_data
```

```python
breast_cancer_data.data
```

```python
breast_cancer_data.feature_names
```

```python
print(breast_cancer_data.DESCR)
```

```python
breast_cancer_data_df=pd.DataFrame(data=breast_cancer_data.data,columns=breast_cancer_data.feature_names)
breast_cancer_data_df["Target"]=breast_cancer_data.target
breast_cancer_data_df
```

## 3. Data understanding

```python
breast_cancer_data_df.shape
```

```python
breast_cancer_data_df.isna().sum()
```

```python
breast_cancer_data_df.dtypes
```

```python
breast_cancer_data_df.describe()
```

## 4. Model Building

```python
X=breast_cancer_data_df.drop(labels="Target",axis=1)
```

```python
X
```

```python
y=breast_cancer_data_df["Target"]
```

```python
y
```

```python
X.shape,y.shape
```

```python
y.value_counts()
```

```python
plt.pie(x=breast_cancer_data_df["Target"].value_counts(),labels=["Benign Cancer","Malignant Cancer"],autopct="%1.0f%%")
plt.show()
```

### 4.1 with stratify=None

```python
# train_test_split with stratify=None
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,shuffle=True,random_state=123,stratify=None)# train_test_split with stratify=None
```

```python
print("Training Samples:",X_train.shape,y_train.shape)
print("Testing Samples:",X_test.shape,y_test.shape)
```

```python
plt.pie(x = y_train.value_counts(),labels = ["Benign Cancer","Malignant Cancer"],autopct = "%1.0f%%")
plt.show()
```

```python
plt.pie(x = y_test.value_counts(),labels = ["Benign Cancer","Malignant Cancer"],autopct = "%1.0f%%")
plt.show()
```

### 4.2 with stratify=y

```python
# train_test_split with stratify=y. it will maintain the proportion of the classes as same as the original dataset in the train and test data too.
X_train_s,X_test_s,y_train_s,y_test_s=train_test_split(X,y,test_size=0.20,shuffle=True,random_state=123,stratify=y)
```

```python
print("Training Samples:",X_train_s.shape,y_train_s.shape)
print("Testing Samples:",X_test_s.shape,y_test_s.shape)
```

```python
plt.pie(x = y_train_s.value_counts(),labels = ["Benign Cancer","Malignant Cancer"],autopct = "%1.0f%%")
plt.show()
```

```python
plt.pie(x = y_test_s.value_counts(),labels = ["Benign Cancer","Malignant Cancer"],autopct = "%1.0f%%")
plt.show()
```

## 5. Model Training

### 5.1 without any Handling imbalance technique

```python
from sklearn.tree import DecisionTreeClassifier
dt_model=DecisionTreeClassifier(criterion='gini',max_depth=3)
dt_model.fit(X_train,y_train)
```

### 5.2 Handing with imbalance technique 1 - stratify

```python
dt_model=DecisionTreeClassifier(criterion='gini',max_depth=3)
dt_model.fit(X_train_s,y_train_s)
```

### 5.3 Handing with imbalance technique 2 - using "class_weight" attribute

```python
from sklearn.tree import DecisionTreeClassifier
dt_model = DecisionTreeClassifier(criterion='gini', max_depth=3,class_weight={0:2,1:1})
dt_model.fit(X_train,y_train)
```

## 6. Model Testing || 7. Model Evaluation

### 6.1 Training Data - without any imbalance technique

```python
y_pred_train=dt_model.predict(X_train)
y_pred_train
```

```python
# Evaluation metrics
from sklearn.metrics import confusion_matrix,classification_report
```

```python
print(classification_report(y_train,y_pred_train))
```

```python
print(confusion_matrix(y_train,y_pred_train))
```

### Test Data

```python
y_pred_test=dt_model.predict(X_test)
y_pred_test
```

```python
print(classification_report(y_test,y_pred_test))
```

```python
print(confusion_matrix(y_test,y_pred_test))
```

### 6.2 Training Data - with imbalance technique 1 - stratify=y

```python
y_pred_train_s=dt_model.predict(X_train_s)
y_pred_train_s
```

```python
print(classification_report(y_train_s,y_pred_train_s))
```

```python
print(confusion_matrix(y_train_s,y_pred_train_s))
```

### Test Data

```python
y_pred_test_s=dt_model.predict(X_test_s)
y_pred_test_s
```

```python
print(classification_report(y_test_s,y_pred_test_s))
```

```python
print(confusion_matrix(y_test_s,y_pred_test_s))
```

### 6.3 Training Data - with imbalance technique 2 - using "class_weight" attribute

```python
y_pred_train_cls=dt_model.predict(X_train)
y_pred_train_cls
```

```python
print(classification_report(y_train,y_pred_train_cls))
```

```python
print(confusion_matrix(y_train,y_pred_train_cls))
```

### Test Data

```python
y_pred_test_cls=dt_model.predict(X_test)
y_pred_test_cls
```

```python
print(classification_report(y_test,y_pred_test_cls))
```

```python
print(confusion_matrix(y_test,y_pred_test_cls))
```
