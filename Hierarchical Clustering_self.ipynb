## 1. Import necessary Libraries

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
import scipy.cluster.hierarchy as sch
from sklearn.cluster import AgglomerativeClustering
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score,precision_score,recall_score,confusion_matrix
from matplotlib import pyplot as plt
```

## 2. Load the data

```python
college_data=pd.read_csv("college_list.csv")
college_data
```

## 4. Data Preparation

### 4.1 Delete unnecessary features

```python
college_data.head()
```

```python
X=college_data.drop("College_Name",axis=1)
X
```

### 4.2 Data transformation - It is a mandatory step to transform the input features before the model building since the algorithm is based on the distance calculation.

```python
scaled_X=StandardScaler()
scaled_X=scaled_X.fit_transform(X)
scaled_X=pd.DataFrame(scaled_X)
scaled_X
```

```python
# To verify whether the mean=0 and standard deviation =1 across all the features
print("Mean:",scaled_X.mean().round(4))
print("Standard Deviation:\n",scaled_X.std().round(4))
```

## 5. Model Building

### Step 1: Let's check DENDOGRAM

```python
plt.figure(figsize=(10,6))
dendogram=sch.dendrogram(sch.linkage(scaled_X,method="ward"))
plt.title("Dendogram for college clustering")
plt.xlabel("Colleges")
plt.ylabel("Eculidean Distance")
plt.show()
```

### For understanding purpose take only 20 colleges to visualize

```python
test_X=scaled_X.head(20)
plt.figure(figsize=(10,6))
dendogram=sch.dendrogram(sch.linkage(test_X,method="ward"))
plt.title("Dendogram for college clustering")
plt.xlabel("Colleges")
plt.ylabel("Eculidean Distance")
plt.show()
```

### Step 2: Apply Agglomerative Clustering(Based on dendrogram, assume 3 clusters)

```python
hc=AgglomerativeClustering(n_clusters=3,metric='euclidean',linkage='ward')
hc
```

```python
clusters=hc.fit_predict(scaled_X)
clusters
```

### Clustering is DONE here. Let's put this back to the original Dataset for further Data Mining Operations
### ===================================================================

### LEVEL 2: Attach Clusters to Dataset

```python
college_data["College_Level"]=clusters
college_data
```

### 2. Data Understanding - Analyze Clusters

```python
college_data[college_data["College_Level"]==0]
```

```python
college_data.groupby(by="College_Level").agg(["mean"])
```

### We did some Mistake here. What is that?
#### We had that categorical feature(College_Name) and we were not able to take mean for the complete Dataset

```python
college_data
```

```python
# will take X which is not having the college_name. we will add the college_level cluster value here
X
```

```python
X["College_Level"]=clusters
X
```

```python
X.groupby(by="College_Level").agg(["mean"])
```

```python
X.groupby(by="College_Level").agg(['min','max'])
```

```python
cluster_names={1:"Best Colleges",
               2:"Mid-Level Colleges",
               0:"Emerging / Lower-Level Colleges"}

X["Category"]=X["College_Level"].map(cluster_names)
X
```

### LEVEL 3: Now it's time to Transition to Supervised Learning

```python
X.head()
```

```python
X_input_data=X.drop(labels=["College_Level","Category"],axis=1)
y_output_data=X["College_Level"]

X_input_data.shape,y_output_data.shape
```

```python
X_train,X_test,y_train,y_test=train_test_split(X_input_data,y_output_data,test_size=0.20,shuffle=True,random_state=42,stratify=y_output_data)
```

```python
rf_model=RandomForestClassifier()
rf_model.fit(X_train,y_train)
# Train Data
y_pred_train=rf_model.predict(X_train)

print("Train Accuracy Score:",accuracy_score(y_train,y_pred_train))

# Test Data
y_pred_test=rf_model.predict(X_test)

print("Test Accuracy Score:",accuracy_score(y_test,y_pred_test))
```
