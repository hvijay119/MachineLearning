## 1. Import necessary Libraries

```python
import pandas as pd
from sklearn.ensemble import GradientBoostingRegressor
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error,mean_squared_error,root_mean_squared_error,r2_score
import pickle as pkl
```

## 2. Load the dataset

```python
california_housing_data=datasets.fetch_california_housing()
california_housing_data_df=pd.DataFrame(data=california_housing_data.data,columns=california_housing_data.feature_names)
california_housing_data_df["target"] = california_housing_data.target
california_housing_data_df
```

## 4. Data Preparation

```python
X=california_housing_data_df.drop(labels="target",axis=1)
```

```python
y=california_housing_data_df["target"]
```

## 5. Model building || 6. Model Training

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,shuffle=True,random_state=42)
```

```python
gradientboosting_model = GradientBoostingRegressor(max_depth=3, n_estimators=50, learning_rate=1.0,)
gradientboosting_model.fit(X_train, y_train)
```

## 7. Model Testing || 8. Model Evaluation

### Training data

```python
y_pred_train = gradientboosting_model.predict(X_train)
```

```python
print("Mean absolute error:",mean_absolute_error(y_train,y_pred_train))
print("Mean squared error:",mean_squared_error(y_train,y_pred_train))
print("Root mean absolute error:",root_mean_squared_error(y_train,y_pred_train))
print("r2 score:",r2_score(y_train,y_pred_train))
```

### Test Data

```python
y_pred_test = gradientboosting_model.predict(X_test)
```

```python
print("Mean absolute error:",mean_absolute_error(y_test,y_pred_test))
print("Mean squared error:",mean_squared_error(y_test,y_pred_test))
print("Root mean absolute error:",root_mean_squared_error(y_test,y_pred_test))
print("r2 score:",r2_score(y_test,y_pred_test))
```

## 9. Model deployment - export the model

```python
pkl.dump(gradientboosting_model,open("gradientboosting_model.pkl","wb"))
```
