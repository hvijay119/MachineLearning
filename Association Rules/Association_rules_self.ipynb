### Let's build BEST ASSOCIATES for Germany as it has more transactions

```python
germany_data=online_retail_data[online_retail_data["Country"]=="Germany"]
germany_data
```

```python
# We need pivot table since apriori algorithim requires the onehot dataframe where it take 0 and 1 as the value for the features
germany_pivoted_data=pd.pivot_table(data=germany_data,values="Quantity",index="InvoiceNo",columns="Description",aggfunc = 'sum').fillna(0)
germany_pivoted_data
```

```python
#sa_data= online_retail_data[online_retail_data["Country"] == "Saudi Arabia"]
#sa_data.head(20)
```

```python
#sa_pivoted_data=pd.pivot_table(data=sa_data,values="Quantity",index="InvoiceNo",columns="Description",aggfunc = 'sum')
#sa_pivoted_data
```

```python
# Now write a lambda funtion to update the values as 1 for the features which containing the value other than 0, so that it will be like onehot DF.
germany_pivoted_data_onehot=germany_pivoted_data.applymap(lambda x:0 if x==0 else 1)
germany_pivoted_data_onehot
```

## 4. Data Mining

```python
from mlxtend.frequent_patterns import apriori,association_rules 
```

```python
# apriori algorithm to find the frequent itemsets
freq_items=apriori(df=germany_pivoted_data_onehot, min_support=0.02,use_colnames=True)
freq_items
```

```python
best_associates=association_rules(df=freq_items,metric='confidence',min_threshold=0.02,)
best_associates
```

```python
best_associates.to_csv("best_anticidents_consequents_germany.csv")
```
